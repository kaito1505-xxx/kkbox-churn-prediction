```python
import pandas as pd
import numpy as np

import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))
```

```python
BASE_DIR = Path("..")
DATA_RAW = BASE_DIR / "data" / "raw"
DATA_INTERIM = BASE_DIR / "data" / "interim"
DATA_PROCESSED = BASE_DIR / "data" / "processed"

ID_COL = "msno"
TARGET_COL = "is_churn"
```

```python
sample_submission = pd.read_csv(DATA_RAW/"sample_submission_v2.csv")
```

```python
sample_submission.head()
```

```python
april_predictions = pd.read_csv(DATA_PROCESSED/"predictions"/"april_2017_predictions.csv")
```

```python
april_predictions.isna().sum()
```

```python
assert len(april_predictions) == len(sample_submission)
assert april_predictions[ID_COL].is_unique
assert april_predictions[ID_COL].equals(sample_submission["msno"])
assert april_predictions[TARGET_COL].notna().all()
assert np.isfinite(april_predictions[TARGET_COL]).all()
assert april_predictions[TARGET_COL].between(0, 1).all()
```
