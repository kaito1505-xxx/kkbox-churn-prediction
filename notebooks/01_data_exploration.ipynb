```python
keys = list(transactions_1.columns)

v2_in_v1 = transactions_2.merge(
    transactions_1[keys].drop_duplicates(),
    on=keys,
    how="inner"
)

print("Rows in transactions_v2:", len(transactions_2))
print("Rows of transactions_v2 found in transactions:", len(v2_in_v1))
print("Rows from transactions_v2 missing from transactions:", len(transactions_2) - len(v2_in_v1))
```

```python
print(transactions_1.dtypes)
print(transactions_2.dtypes)
```

```python
for name, df in [("v1", transactions_1), ("v2", transactions_2)]:
    print(name)
    print("dates:", df["transaction_date"].min(), df["transaction_date"].max())
    print("users:", df["msno"].nunique())
    print("duplicate full rows:", df.duplicated().sum())
```

```python
t1_users = set(transactions_1["msno"].dropna().unique())
t2_users = set(transactions_2["msno"].dropna().unique())
train1_users = set(train_1["msno"].dropna().unique())
train2_users = set(train_2["msno"].dropna().unique())

for name, cohort in [("train1", train1_users), ("train2", train2_users)]:
    print(name, "users:", len(cohort))
    print("in transactions_v1:", len(cohort & t1_users))
    print("in transactions_v2:", len(cohort & t2_users))
    print("only in v1:", len(cohort & (t1_users - t2_users)))
    print("only in v2:", len(cohort & (t2_users - t1_users)))
    print("not in a single one:", len(cohort - (t1_users | t2_users)))

print("Transaction users in both versions:", len(t1_users & t2_users))
```

```python
for path in [DATA_RAW/"user_logs.csv", DATA_RAW/"user_logs_v2.csv"]:
    print(pd.read_csv(path, nrows=5).dtypes)
```

```python
log_cols = [
    "msno", "date", "num_25", "num_50", "num_75",
    "num_985", "num_100", "num_unq", "total_secs"
]

for path in [DATA_RAW/"user_logs.csv", DATA_RAW/"user_logs_v2.csv"]:
    rows = 0
    min_date = None
    max_date = None
    users = set()
    null_counts = pd.Series(dtype="int64")

    for chunk in pd.read_csv(
        path,
        usecols=log_cols,
        chunksize=500_000,
        dtype={"msno": "string", "date": "Int64"}
    ):
        rows += len(chunk)

        chunk_min = chunk["date"].min()
        chunk_max = chunk["date"].max()
        min_date = chunk_min if min_date is None else min(min_date, chunk_min)
        max_date = chunk_max if max_date is None else max(max_date, chunk_max)

        users.update(chunk["msno"].dropna().unique())
        null_counts = null_counts.add(chunk.isna().sum(), fill_value=0)

    print(path)
    print("rows:", rows)
    print("users:", len(users))
    print("date range:", min_date, max_date)
    print("missing values ​​by column:\n", null_counts.astype("int64"))
```

```python
transactions = pd.concat(
    [transactions_1, transactions_2],
    ignore_index=True
).drop_duplicates()
```

```python
import pyarrow as pa
import pyarrow.parquet as pq

paths = [
    DATA_RAW/"user_logs.csv",
    DATA_RAW/"user_logs_v2.csv",
]
output = "../data/interim/user_logs.parquet"

dtypes = {
    "msno": "string",
    "date": "Int64",
    "num_25": "Int64",
    "num_50": "Int64",
    "num_75": "Int64",
    "num_985": "Int64",
    "num_100": "Int64",
    "num_unq": "Int64",
    "total_secs": "float64",
}

writer = None

try:
    for path in paths:
        for chunk in pd.read_csv(path, chunksize=500_000, dtype=dtypes):
            table = pa.Table.from_pandas(chunk, preserve_index=False)

            if writer is None:
                writer = pq.ParquetWriter(
                    output, table.schema, compression="snappy"
                )

            writer.write_table(table)
finally:
    if writer is not None:
        writer.close()
```
